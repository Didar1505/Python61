```python
def calculate_total(price, quantity):
    return price * quantity

def apply_discount(total, percent):
    return total - total * percent / 100

total = calculate_total(500, 4)
final_price = apply_discount(total, 10)
print("Стоимость товаров:", total)
print("Стоимость со скидкой:", final_price)
```

## 12. Решение мини-проекта

Для преподавателя и самостоятельной проверки после выполнения всех этапов.

```python
def calculate_total(price, quantity):
    return price * quantity


def get_discount(total):
    if total >= 5000:
        return 10
    elif total >= 2000:
        return 5
    else:
        return 0


def apply_discount(total, percent):
    discount = total * percent / 100
    final_price = total - discount
    return discount, final_price


def calculate_delivery(total):
    if total >= 3000:
        return 0
    return 300


def can_pay(balance, total):
    return balance >= total


def calculate_order(price, quantity):
    total = calculate_total(price, quantity)
    percent = get_discount(total)
    discount, final_price = apply_discount(total, percent)
    delivery = calculate_delivery(final_price)
    order_total = final_price + delivery
    return total, percent, discount, final_price, delivery, order_total
```

```python
price = 1200
quantity = 3
balance = 5000

# При необходимости используйте ввод с клавиатуры:
# price = float(input("Цена товара: "))
# quantity = int(input("Количество товаров: "))
# balance = float(input("Ваш баланс: "))

total, percent, discount, final_price, delivery, order_total = calculate_order(price, quantity)

print("========== ВАШ ЗАКАЗ ==========")
print("Стоимость товаров:", total, "руб.")
print("Скидка:", percent, "%")
print("Размер скидки:", discount, "руб.")
print("Стоимость со скидкой:", final_price, "руб.")
print("Доставка:", delivery, "руб.")
print("Итого к оплате:", order_total, "руб.")

if can_pay(balance, order_total):
    print("Оплата доступна.")
    print("Остаток на балансе:", balance - order_total, "руб.")
else:
    print("Недостаточно средств.")
    print("Не хватает:", order_total - balance, "руб.")

print("==============================")
```

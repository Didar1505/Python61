## 1. Локальные переменные

**Локальная переменная** — имя, которому присваивается значение внутри функции. Она доступна в этой функции. Параметры также являются локальными именами.

### 1.1. Простейший пример

Переменная `name` создаётся при вызове функции.

````python
def say_hello():
    name = "Алексей"
    print("Привет,", name)

say_hello()
````

### 1.2. Стоимость покупки

Промежуточные расчёты удобно хранить в локальных переменных.

````python
def calculate_total():
    price = 500
    quantity = 3
    total = price * quantity
    print("Стоимость покупки:", total)

calculate_total()
````

### 1.3. Параметры и return

`price`, `quantity` и `total` — локальные имена. `return` возвращает значение, а не само имя.

````python
def calculate_total(price, quantity):
    total = price * quantity
    return total

result = calculate_total(250, 4)
print("Итого:", result)
````

### 1.4. Переменные разных функций

Одинаковые локальные имена в разных функциях не мешают друг другу.

````python
def calculate_salary():
    money = 30000
    print("Зарплата:", money)

def calculate_bonus():
    money = 5000
    print("Премия:", money)

calculate_salary()
calculate_bonus()
````

## 2. Область видимости локальной переменной

**Область видимости** — часть программы, в которой доступно имя переменной. Локальное имя не становится доступным снаружи после вызова функции.

### 2.1. Доступ внутри и снаружи функции

Снимите комментарий с последней строки: возникнет `NameError`, поскольку имя `age` снаружи не определено.

````python
def show_age():
    age = 18
    print(age)

show_age()
# print(age)  # NameError
````

### 2.2. Получение значения через return

Значение можно сохранить в другой переменной за пределами функции.

````python
def calculate_sum():
    a = 10
    b = 20
    result = a + b
    return result

total = calculate_sum()
print(total)
````

### 2.3. Одинаковые имена

Локальное имя `number` скрывает одноимённое глобальное имя внутри функции.

````python
number = 100

def show_number():
    number = 20
    print("Внутри функции:", number)

show_number()
print("Снаружи функции:", number)
````

### 2.4. Использование до присваивания

Если прочитать локальную переменную до присваивания, возникнет `UnboundLocalError`. Для проверки снимите комментарий с вызова.

````python
def show_price():
    print(price)
    price = 500

# show_price()  # UnboundLocalError
````

### 2.5. Исправление

Сначала присваиваем значение, затем читаем его.

````python
def show_price():
    price = 500
    print(price)

show_price()
````

## 3. Глобальные переменные

**Глобальная переменная** — имя на уровне модуля (в данном случае — основной программы). Функции могут читать её, если не скрывают имя локальной переменной.

### 3.1. Чтение глобальной переменной

Для чтения `global` не нужен.

````python
message = "Добро пожаловать!"

def show_message():
    print(message)

show_message()
````

### 3.2. Общие настройки приложения

Несколько функций используют одно значение.

````python
app_name = "Мой магазин"

def show_header():
    print("Приложение:", app_name)

def show_welcome():
    print("Добро пожаловать в", app_name)

show_header()
show_welcome()
````

### 3.3. Совпадение имён

Присваивание внутри функции без `global` создаёт локальное имя, глобальное значение не меняется.

````python
balance = 1000

def show_balance():
    balance = 500
    print("Внутри:", balance)

show_balance()
print("Снаружи:", balance)
````

### 3.4. Ключевое слово global

`global` позволяет присваивать новое значение глобальному имени внутри функции.

````python
counter = 0

def increase_counter():
    global counter
    counter = counter + 1

increase_counter()
increase_counter()
increase_counter()
print("Счётчик:", counter)
````

### 3.5. Счётчик посетителей

Каждый вызов изменяет общее значение.

````python
visitors = 0

def register_visitor():
    global visitors
    visitors = visitors + 1
    print("Посетителей:", visitors)

register_visitor()
register_visitor()
register_visitor()
````

### 3.6. Типичная ошибка

Из-за присваивания Python считает `counter` локальным во всей функции. Чтение справа происходит до присваивания. Снимите комментарий для проверки.

````python
counter = 0

def increase():
    counter = counter + 1

# increase()  # UnboundLocalError
````

### 3.7. Исправление через global

Функция явно изменяет глобальное имя.

````python
counter = 0

def increase():
    global counter
    counter = counter + 1

increase()
print(counter)
````

### 3.8. Предпочтительный вариант: параметры и return

Функция не зависит от глобального счётчика. Результат сохраняется снаружи.

````python
def increase(number):
    return number + 1

counter = 10
counter = increase(counter)
print(counter)
print(increase(5))
print(increase(100))
````

### Сравнение

| Признак | Локальная | Глобальная |
|---|---|---|
| Где создаётся | В функции | На уровне модуля |
| Где доступна | В своей функции | В модуле и в функциях, если имя не скрыто |
| Присваивание внутри функции | Обычное | Для глобального имени нужен `global` |
| Назначение | Промежуточные вычисления | Общие данные |

**Рекомендация:** чаще используйте параметры и `return`. Не изменяйте глобальные переменные без необходимости.

## 4. Глобальные константы

**Константа** — значение, которое по замыслу программы не меняется. В Python имена констант принято писать заглавными буквами, разделяя слова подчёркиваниями (`MAX_ATTEMPTS`). Это соглашение, а не ограничение языка.

### 4.1. Скидка интернет-магазина

Общая ставка скидки хранится в одном месте.

````python
DISCOUNT = 0.10

def calculate_price(price):
    discount_amount = price * DISCOUNT
    return price - discount_amount

print(calculate_price(1000))
````

### 4.2. Перевод часов в минуты

Константа объясняет смысл числа 60.

````python
MINUTES_IN_HOUR = 60

def convert_to_minutes(hours):
    minutes = hours * MINUTES_IN_HOUR
    return minutes

print(convert_to_minutes(2))
print(convert_to_minutes(5))
````

### 4.3. Настройка приложения

Используем именованное постоянное значение.

````python
MAX_ATTEMPTS = 3

def show_limit():
    print("Максимум попыток:", MAX_ATTEMPTS)

show_limit()
````

### 4.4. Можно ли изменить константу?

Python разрешает повторное присваивание, но оно нарушает соглашение о константах.

````python
MAX_ATTEMPTS = 3
print(MAX_ATTEMPTS)

MAX_ATTEMPTS = 5  # Технически допустимо, но не рекомендуется
print(MAX_ATTEMPTS)
````

| Признак | Глобальная переменная | Глобальная константа |
|---|---|---|
| Имя | `user_count` | `MAX_USERS` |
| Изменение по замыслу | Допускается | Не предполагается |
| Python запрещает присваивание | Нет | Нет |

### Решение 17

Расчёт использует параметры, локальные переменные и константы; регистрация изменяет только счётчик.

````python
DISCOUNT = 0.10
DELIVERY_PRICE = 200
orders_count = 0

def calculate_order(price, quantity):
    total = price * quantity
    discount_amount = total * DISCOUNT
    total = total - discount_amount
    total = total + DELIVERY_PRICE
    return total

def register_order():
    global orders_count
    orders_count = orders_count + 1

first_order = calculate_order(1000, 2)
register_order()

second_order = calculate_order(500, 3)
register_order()

print("Первый заказ:", first_order)
print("Второй заказ:", second_order)
print("Всего заказов:", orders_count)
````
